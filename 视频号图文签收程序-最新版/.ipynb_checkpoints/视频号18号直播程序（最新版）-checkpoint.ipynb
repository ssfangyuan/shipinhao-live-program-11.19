```python
import pandas as pd
df =pd.read_excel("./1.xlsx")
```

```python
keywords=['10000222041883']
df['商品编码(平台)'].isin(keywords)
df['商品编码(平台)'] = df['商品编码(平台)'].astype(str)
#删除其他行
df.drop(index= df[ df['商品编码(平台)'].isin(keywords)==False].index,inplace=True)
df.reset_index(drop=True,inplace= True)
```

```python
df.rename(columns={"订单下单时间":"下单时间"},inplace=True)
```

```python
df
```

```python
if '\t' in df.loc[0,'下单时间']:
    df['下单时间'] = df['下单时间'].str[1:]
```

```python
from datetime import datetime
df["下单时间"] = pd.to_datetime(df["下单时间"],format = "%Y-%m-%d %H:%M:%S")
df["下单时间"] = df["下单时间"].map(lambda x: x.strftime('%Y-%m-%d'))
df["销售金额"]= (df['商品数量'] * df['商品价格']) + (df['订单运费'])
df1 = pd.read_excel("商品编码.xlsx")
df1.rename(columns={"商品名":"产品名称","商家编码":"SKU编码(自定义)"},inplace=True)
df1 = df1[["SKU编码(自定义)","产品名称"]]
df= pd.merge(df,df1,on="SKU编码(自定义)",how="left")
```

```python
# 缺码表
lackCode = df[df['产品名称'].isnull()]['SKU编码(自定义)'].value_counts()
lackCode
```

```python
lackCode.to_excel('缺码表.xlsx')
```

```python
df['订单数'] = df['商品数量']
```

```python
from datetime import datetime
#df["支付时间"] = df["支付时间"].fillna(pd.to_datetime('1970-01-01'))
df["支付时间"] = pd.to_datetime(df["支付时间"],format = "%Y-%m-%d %H:%M:%S",errors='coerce')
df["支付时间"] = df["支付时间"].fillna(pd.to_datetime('1970-01-01'))
df["支付时间"] = df["支付时间"].map(lambda x: x.strftime('%Y-%m-%d'))
```

```python
df["下单单量"] = df['订单数']
df["支付单量"] = 0
df["发货单量"] = 0
df["发货前退款单量"] = 0
```

```python
for i in range(0,len(df)):
    if df.loc[i,"支付时间"] != "1970-01-01":
        df.loc[i,"支付单量"] = df.loc[i,'订单数']
```

```python
for i in range(0,len(df)):
    if df.loc[i,"商品发货"] == "已发货" :
        df.loc[i,"发货单量"] = df.loc[i,'订单数']
```

```python
for i in range(0,len(df)):
    if df.loc[i,"支付时间"] != "1970-01-01" and  df.loc[i,"商品发货"] == "未发货" and df.loc[i,"商品售后"] == "退款完成":
        df.loc[i,"发货前退款单量"] = df.loc[i,'订单数']
```

```python
df["已签收"] = 0
df["发货中1"] = 0
df["售后中"] = 0
df["已退款"] = 0
df["预计签收单量"] = 0
df["发货中2"] = 0
for i in range(0,len(df)):
    if df.loc[i,"订单状态"] == "已完成" and ((df.loc[i,'商品售后'] == "无") or (df.loc[i,'商品售后'] == "用户取消申请") or (df.loc[i,'商品售后'] == "退货退款关闭")):
         df.loc[i,"已签收"] = df.loc[i,'订单数']
    elif df.loc[i,"订单状态"] == "已发货" and ((df.loc[i,'商品售后'] == "无") or (df.loc[i,'商品售后'] == "用户取消申请") or (df.loc[i,'商品售后'] == "退货退款关闭") ):
         df.loc[i,"发货中1"] = df.loc[i,'订单数']
    elif df.loc[i,'商品售后'] =="待买家退货" or df.loc[i,'商品售后'] =="待商家收货" or df.loc[i,'商品售后'] =="客服介入中" or df.loc[i,'商品售后'] =="商家受理中" :
        df.loc[i,"售后中"] = df.loc[i,'订单数']
for i in range(0,len(df)):
    if df.loc[i,"商品发货"] == "已发货" and (df.loc[i,'商品售后'] == "退货退款完成" or df.loc[i,'商品售后'] == "退款完成"):
        df.loc[i,"已退款"] = df.loc[i,'订单数']
for i in range(0,len(df)):
    if df.loc[i,"商品发货"] == "已发货" and (df.loc[i,'商品售后'] == "无" or df.loc[i,'商品售后'] == "用户取消申请" or df.loc[i,'商品售后'] == "退货退款关闭"):
         df.loc[i,"预计签收单量"] = df.loc[i,'订单数']
for i in range(0,len(df)):
    if df.loc[i,"订单状态"] == "已取消" and  df.loc[i,"商品发货"] == "已发货" and (df.loc[i,'商品售后'] == "用户取消申请" or df.loc[i,'商品售后'] == "退货退款关闭" ) :
        df.loc[i,"发货中2"] = df.loc[i,'订单数']
```

```python
import numpy as np
df['销售金额'].fillna(0, inplace=True)
result = pd.pivot_table(df,index=['下单时间',"产品名称"] , values = ['下单单量','支付单量','发货单量','已签收','发货中1','发货中2','售后中','已退款','发货前退款单量',"预计签收单量"] , aggfunc = np.sum)
result = result.reset_index()
result = result[['下单时间',"产品名称",'下单单量','支付单量','发货单量','已签收','发货中1','发货中2','售后中','已退款','发货前退款单量',"预计签收单量"]]
```

```python
#拼接店铺消耗表
df_consume = pd.read_excel("店铺消耗微信豆25号.xlsx")
df_consume = df_consume[['日期','产品名称','累计消耗金额','成交订单数']]
temp = df_consume.日期.str.extract('(\d+).*?(\d+).*?(\d+)') 
df_consume['下单时间'] = temp[0] + "-" + temp[1] + '-' + temp[2]
del df_consume['日期']
df_consume = pd.pivot_table(df_consume,index=['下单时间','产品名称'] , values = ['累计消耗金额','成交订单数'] , aggfunc = np.sum)
df_consume = df_consume.reset_index()
df_consume["下单时间"] = pd.to_datetime(df_consume["下单时间"],format = "%Y-%m-%d")
df_consume["下单时间"] = df_consume["下单时间"].map(lambda x: x.strftime('%Y-%m-%d'))
```

```python
#拼接店铺消耗表
df_consume1 = pd.read_excel("店铺消耗ADQ25号.xlsx")
df_consume1 = df_consume1[['日期','产品名称','累计消耗金额','成交订单数']]
temp1 = df_consume1.日期.str.extract('(\d+).*?(\d+).*?(\d+)') 
df_consume1['下单时间'] = temp1[0] + "-" + temp[1] + '-' + temp[2]
del df_consume1['日期']
df_consume1 = pd.pivot_table(df_consume1,index=['下单时间','产品名称'] , values = ['累计消耗金额','成交订单数'] , aggfunc = np.sum)
df_consume1 = df_consume1.reset_index()
df_consume1["下单时间"] = pd.to_datetime(df_consume1["下单时间"],format = "%Y-%m-%d")
df_consume1["下单时间"] = df_consume1["下单时间"].map(lambda x: x.strftime('%Y-%m-%d'))
df_consume1.rename(columns={'成交订单数': 'ADQ成交订单数','累计消耗金额': 'ADQ累计消耗金额'}, inplace=True)

print(df_consume1)
```

```python
final = pd.merge(result,df_consume,on = ['下单时间','产品名称'],how = 'left')

final = final[['下单时间','产品名称','成交订单数','下单单量','支付单量','发货单量','已签收','发货中1','发货中2','售后中','已退款','发货前退款单量',"预计签收单量","累计消耗金额"]]
final['发货中'] = final['发货中1'] + final['发货中2']
final = final[['下单时间','产品名称','成交订单数','下单单量','支付单量','发货单量','已签收','发货中','售后中','已退款','发货前退款单量',"预计签收单量","累计消耗金额"]]
final.rename(columns={"成交订单数":"广告单量"},inplace=True)

final.rename(columns={'累计消耗金额': '微信豆累计消耗金额'}, inplace=True)

final = pd.merge(final,df_consume1,on = ['下单时间','产品名称'],how = 'left')
```
